### Analysis IV - Day of Week & Ozone Levels
It was theorized that ozone levels may differ between weekdays & weekends because different kinds & quantities of urban activity usually take place during the week & on the weekend. For example, a large chunk of the general population has to commute to & from their workplace between Monday & Friday, whereas they are more likely to stay home on the weekend. Through the usage of transportation, gas & diesel vehicles emit chemicals & gases that can form into ozone in certain conditions, specifically when interacting with sunlight & high temperatures. This reasoning is behind the inquiry as to whether urban activity might relate to ozone levels.

In _Cleaning Data & the Skies I_, straightforward statistical analyses were conducted the results of which proved inconclusive. More specifically, ozone levels were analyzed across weekdays vs weekends, different days of the week, on a monthly basis, & across a specific county. Ultimately, they failed to find any meaningful differences in the ozone levels.  
This file, _Cleaning Data & the Skies II_, extends these analyses into more powerful statistical techniques, specifically hypothesis tests.

Given the nature of the data, parametric, two-tailed t-tests were used to evaluate whether average ozone levels were different between weekdays & weekends in California in 2024. First, these averages were assessed across the entire dataset. Using a significance level of five percent, the hypothesis test yielded a p-value of about 0.26%. As such, the null hypothesis was rejected in favor of the alternative. More specifically, the average ozone level observed during the week was about 0.00043 ppm higher than on the weekends. The p-value indicates that this difference is statistically significant.

Secondly, average ozone levels across weekdays & weekends were evaluated specifically for the county of Los Angeles. Using a significance level of five percent, the hypothesis test yielded a p-value of about 4.06%. As such, the null hypothesis was rejected in favor of the alternative. More specifically, the average ozone level observed on the weekends in Los Angeles county was about 0.001143 ppm higher than during the week. The p-value indicates that this difference is statistically significant.

__Recommendations__: If it is assumed that urban activity is the main catalyst regarding differences in ozone levels between weekends & weekdays in this dataset, then these results indicate that urban activity did have a meaningful effect on ozone levels observed in California in 2024 across different days of the week. With this in mind, urban activity generally resulted in ozone levels being slightly higher during the week than on weekends during 2024 in California. Conversely, for the county of Los Angeles, urban activity resulted in ozone levels being slightly lower during the week than on weekends during 2024.

This process in determining what dates are interpretable or not will form the basis for cleaning these "/2024" dates moving forward. So, for each observation whose 'Date' is "/2024," the following need to be obtained.
- The most previous non-missing date & how many rows prior it is.
- The immediately subsequent non-missing date & how many rows after it is.

These quantities can then be used to determine if a vague date ("/2024") is "interpretable" or not. Then the "interpretable" dates can be interpolated accordingly.

When undergoing this process, some of the vague dates are directly adjacent to observations that also have vague dates. This would necessitate muliple "for" loops when determining the closest preceding & subsequent data point with a real date. However, it is necessary to know how many "for" loops are required. In other words, with how the original dataset is sorted, what is the longest streak of consecutive observations whose 'Date's were missing ("/2024"). This streak will dictate how many "for" loops are necessary when looking backwards or forwards in the dataset. A custom function was created in order to find this longest streak of consecutive "/2024" dates.

```python
#Create a new dataframe focusing on the date variables
date_testing = df_I_A[['Orig Date','Date']].copy()
```

```python
#Want to find the longest streak of consecutive "/2024" 'Date's

def longest_consecutive_dates(nums):
    '''
    A function was created to find the longest streak of consecutive observations which had a 'Date' of "/2024."
    The inserted data should be a list of relevant indices.
    This function will be used for the loop in the following cells which looks for non-missing dates of most adjacent 
    data points.
    '''
    #initialize a list containing the consecutive elements of the longest steak
    longest_streak = []
    #retrieve the current index being looked at
    current_streak = [nums[0]]
    #loop through the indices 1 by 1
    for i in range(1, len(nums)):
        #look for consecutive indices
        if nums[i] == nums[i - 1] + 1:
            #add the current element of the streak
            current_streak.append(nums[i])
        #once the consecutive streak ends
        else:
            #check if the current streak is longer than the longest streak observed so far
            if len(current_streak) > len(longest_streak):
                #if so, adjust the current longest streak
                longest_streak = current_streak
            #retrieve the indices of the longest streak
            current_streak = [nums[i]]

    #check if the current streak is longer than the longest streak observed
    if len(current_streak) > len(longest_streak):
        #if so, adjust the current longest streak
        longest_streak = current_streak
    #return the elements of the longest streak observed
    return longest_streak


#retrieve the result with the longest # consecutive elements w/missing dates & their indices
longest_consecutive_indices_nulldates = longest_consecutive_dates(orig_data[orig_data['Date'] == '/2024'].index.tolist())
display("Longest streak of consecutive indices: {}".format(len(longest_consecutive_indices_nulldates)))
display("Associated indices in original dataset: {}".format(longest_consecutive_indices_nulldates))

#retrieve the associated data
display(orig_data.iloc[12428:12436,:4])
```

#### Interpreting the missing dates
Now that the necessary variables have been constructed, they can be used to determine if the vague dates ("/2024") are potentially interpretable or not. This is essentially a form of interpolation. The steps to create the code are listed below.
1) For each row, check whether the 'Date' is vague ("/2024") or not.

2) Determine if missing dates are interpretable or not using the four new variables.
    - Find the difference (in days) between the 'Most Subsequent Actual Date' & 'Most Previous Actual Date', & then sum the '# Indices Prior' & '# Indices After'. If these two values are equal, then the missing 'Date' is interpretable. If these numbers are NOT equal, then the missing 'Date' is not interpretable, & it should be left alone as pd.NaT (a null timestamp).
        - E.g. In indices 12460:12470, the two consecutive missing dates ARE interpretable because there are two missing values & two dates in between the most adjacent non-missing dates of 2024-10-04 & 2024-10-07 (10-05, 10-06).
        - E.g. In indices 12428:12420, the six consecutive missing dates ARE NOT interpretable because there are six missing values but seven dates in between the most adjacent non-missing dates of 2024-08-29 & 2024-09-06 (08-30, 08-31, 09-01, 09-02, 09-03, 09-04, 09-05).
    - Also, a special case needs to be examined when dates in the dataset flip from the end of the year (e.g. 12-31) back to the beginning of the year (e.g. 01-01). The arithmetic needs to be a little more sophisticated because pandas will naturally think that the date following 2024-12-31 is 2025-01-01; however, in this case, the dates are all in 2024 & so the code will think the difference between dates of 12-31 & 01-01 is a full year instead of a single day. This can be solved by adding a pandas.DateOffset of one year so that the arithmetic doesn't think that a whole year differentiates such dates. This is illustrated in the code below.

Additionally, the very first data point in the dataset has a 'Date' of "/2024," but based on the following values, it is reasonable to presume that it is 2024-01-01. This needs to be directly indicated in the code.

```python
#Create another variable indicating whether a missing 'Date' is interpretable or not using the logic above
date_testing['Missing Date Interpretable?'] = '-'

#retrieve indices with missing 'Date's ("/2024")
for i in [row for row in date_testing.index if pd.isna(date_testing.loc[row,'Date'])]:
    #need to consider the special case at index 0, which is easily self-interpretable given the following 'Date's (2024-01-01)
    if i == 0:
        date_testing.loc[i,'Missing Date Interpretable?'] = 'Yes'
        
    #for the rest of the indices
    else:
        #determine if the missing 'Date' is interpretable or not using appropriate mathematical logic
        if (date_testing.loc[i,'Most Subsequent Actual Date'] - date_testing.loc[i,'Most Previous Actual Date']).days == (
            date_testing.loc[i,'# Indices Prior'] + date_testing.loc[i,'# Indices After']):
            #if these values are equal, the missing 'Date's are interpretable
            date_testing.loc[i,'Missing Date Interpretable?'] = 'Yes'
        
        #consider the special case when the dates changed from the end of the year (e.g. 12-31) back to the beginning (01-01). 
            #adapt the previous logic to get the (chronological) difference in dates (in days) irrespective of year
        elif ((date_testing.loc[i,'Most Subsequent Actual Date'] + pd.DateOffset(years=1)) - date_testing.loc[i,
            'Most Previous Actual Date']).days == (
            date_testing.loc[i,'# Indices Prior'] + date_testing.loc[i,'# Indices After']):
            #if these values are equal, the missing 'Date's are interpretable
            date_testing.loc[i,'Missing Date Interpretable?'] = 'Yes'
        
        else:
            #otherwise, the missing 'Date's are not interpretable
            date_testing.loc[i,'Missing Date Interpretable?'] = 'No'

#check code in a different cell -- GOOD
```

#### Imputing Interpretable Missing Dates
The next step involves utilizing this data to interpolate missing 'Date's that are deemed interpretable. Then, the new dates can be inserted into the main dataset. To do this, `pandas.combine_first()` can be used which imputes non-missing data points from a second dataset into a first dataset only if they are missing in the first one.

```python
#Assign the missing dates that are deemed interpretable. Create another variable indicating the interpolated 'Date'.
date_testing['Interpolated Date'] = pd.NaT

#loop through the relevant data points
for i in [row for row in date_testing.index if date_testing.loc[row,'Missing Date Interpretable?'] == 'Yes']:
    
    #consider special case for the first data point which is reasonably interpretable based on the subsequent dates
    if i == 0:
        date_testing.loc[i,'Interpolated Date'] = pd.to_datetime("2024-01-01")
    
    #for all other indices
    else:
        
        #consider special cases where date changes from end of year to beginning
        if date_testing.loc[i,'Most Previous Actual Date'] > date_testing.loc[i,'Most Subsequent Actual Date']:
            date_testing.loc[i,'Interpolated Date'] = date_testing.loc[i,'Most Previous Actual Date'] + pd.DateOffset(
                years=-1, days=date_testing.loc[i,'# Indices Prior'])
        
        #for other indices
        else:
            date_testing.loc[i,'Interpolated Date'] = date_testing.loc[i,'Most Previous Actual Date'] + pd.DateOffset(
                days=date_testing.loc[i,'# Indices Prior'])

#Check -- GOOD
#date_testing.iloc[:,-1].value_counts().sum() == 8,663
#date_testing.iloc[12428:12470,:]
```

As such, a more powerful statistical method that could discern if any differences do exist may be necessary.


Hypothesis testing is a statistical procedure that takes a sample of a population dataset & evaluates the likelihood that the observed results (in the sample data) occurred due to chance under the assumption that a null hypothesis is true--that there is no difference/relationship between the groups/variables. If the probability that such observations occurred due to chance is sufficiently small, which can be quantified with the p-value, then the null hypothesis can be rejected in favor of the alternative hypothesis--that there is a difference/relationship. The p-value describes the likelihood of observing data (in the population dataset) at least as extreme as what was seen in the sample assuming that the null hypothesis is true. In other words, it quantifies the probability of extreme results occurring due to chance.
- Generally, if the calculated p-value is less than 0.05, than the null hypothesis can be rejected in favor of the alternative. In this case, the p-value describes the probability of observing a difference in ozone levels (in the population dataset) that's at least as extreme as what was observed in the sample data assuming that the null hypothesis is true--that there is not a meaningful difference in ozone levels across weekdays & weekends.
- While hypothesis testing doesn't prove or disprove anything, it is useful when quantifying if there is sufficient evidence to claim that a difference, as stated by an alternative hypothesis, exists.

There are several steps involved in such a process that can dictate what type of hypothesis test is most appropriate. Moreover, there are several aspects of the data that determine this such as the independence of data points, the sample size, how many groups are being compared, the type of data being evaluated, & more. To answer some of these questions, it is also useful to evaluate the distribution of ozone measurements across the two groups (weekdays versus weekends).

Firstly, it is important to establish that the data points being tested are independent. Recall that this section is interested in the differences of ozone levels on a statewide level during the week & on the weekends. Provided the background of this project, it is assumed that data points were measured individually & via the same process regardless of the day or the collection method. There is no indication or reason that indicates that the recording of one measurement was dependent or related to that of another. Given this, it can be assumed that the ozone levels observed during weekdays & those observed during weekends are independent.

```python
#Examine the distribution of ozone levels between Week/Weekends. Add the median, mode ozone level.

fig, axes = plt.subplots(nrows=2, sharex=True)
#WEEK
ax1 = sns.histplot(data=data_IV[data_IV['Week/Weekend'] == 'Week'], x='Daily Max 8-hour Ozone Concentration', ax=axes[0])
#mean
ax1.axvline(x=data_IV[data_IV['Week/Weekend'] == 'Week']['Daily Max 8-hour Ozone Concentration'].mean(), color='red',
            alpha=0.6, linestyle='--')
#median
ax1.axvline(x=data_IV[data_IV['Week/Weekend'] == 'Week']['Daily Max 8-hour Ozone Concentration'].median(), color='orange',
            alpha=0.7, linestyle='--')

#WEEKEND
ax2 = sns.histplot(data=data_IV[data_IV['Week/Weekend'] == 'Weekend'], x='Daily Max 8-hour Ozone Concentration', ax=axes[1])
#mean
ax2.axvline(x=data_IV[data_IV['Week/Weekend'] == 'Week']['Daily Max 8-hour Ozone Concentration'].mean(), color='red',
            alpha=0.6, linestyle='--')
#median
ax2.axvline(x=data_IV[data_IV['Week/Weekend'] == 'Weekend']['Daily Max 8-hour Ozone Concentration'].median(), color='orange',
            alpha=0.7, linestyle='--')

#edit plot details
for k, v in {'Ozone Levels During the Week - CA, 2024':[0], 
            'Ozone Levels During the Weekend - CA, 2024':[1]}.items():
    plt.sca(axes[v[0]])
    plt.title(k, fontsize=10)
    plt.xticks(fontsize=9, ticks=np.arange(0,0.15,0.02))
    #add a legend
    red_line = mlines.Line2D([], [], color='red', linestyle='--', alpha=0.7, label='Mean')
    orange_line = mlines.Line2D([], [], color='orange', linestyle='--', alpha=0.7, label='Median')
    plt.legend(handles=[red_line, orange_line], fontsize=8, title='Ozone Level', title_fontsize=9)
    plt.grid(axis='y',alpha=0.2,color='gray')
    plt.tight_layout();
```

The figures above display the distribution of daily, maximum ozone levels observed across California in 2024. The upper plot contains data observed during the week, whereas the bottom plot contains data observed on the weekends. The red & orange dashed lines indicate the average & median ozone level of each group respectively. Note that ozone levels are in units of parts per million (ppm). The vertical axes of the two plots are **not** identical given that there are about two-and-a-half as many data points observed during the week than there were during the weekend.

Both distributions are quite similar. For both groups, their ozone levels appear to be slightly right skewed; however, there is a clear single peak in both distributions. Moreover, the average ozone levels in the two groups were higher than their respective median ozone levels. Generally, these two metrics across the two groups are very similar--slightly above 0.04 ppm.

For the purposes of hypothesis testing, these histograms indicate that the ozone levels across the two groups--weekdays & weekends--follow a clear distribution. While it is not a perfectly normal distribution, they are skewed very slightly. In other words, the difference between their mean & median ozone levels are very small.  
Moreover, both groups have a significant number of data points so sample size won't be problematic. The Central Limit Theorem states that for a significantly large sample--typically more than 30 data points--the sample means begin to approximate a normal distribution. There are over 14,000 & 36,000 data points measured in the weekend & weekday groups respectively which is more than sufficient for the Central Limit Theorem to apply. Given this & the relatively small difference between mean & median ozone levels for each of the two groups, it is reasonable to assert that these data points can be modeled by an approximately normal distribution at the very least.
> As such, parametric hypothesis tests are appropriate because, while they typically assume normally distributed group means, they can tolerate a slight skew for significantly large sample sizes. Non-parametric tests can still apply, but when the data model a clear distribution like in this case, parametric tests are more powerful.

With this established, other aspects of the dataset need to be clarified. Recall that this section seeks to determine if there is an effect on ozone levels (a continuous variable) based on whether they were measured during the week or on the weekend (two groups). As such, a two-tailed test is appropriate because the test is evaluating a general difference between the two group means. Earlier, it was clarified that the ozone levels between the two groups are independent which reinforces the Central Limit Theorem & makes a parametric t-test applicable.
> As such, a parametric, two-tailed t-test can be used to evaluate whether the average ozone levels observed on weekdays & on weekends, which are independent, are meaningfully different or not. Such a test evaluates whether there is a difference in means between the two groups.

The `scipy.stats.ttest_ind()` function was used for these purposes.

For clarity, the null (_$H_0$_) & alternative (_$H_A$_) hypotheses are listed below.
- _$H_0$_: There is no difference in the ozone levels between weekdays & weekends.
- _$H_A$_: There is a meaningful difference in the ozone levels between weekdays & weekends.
- Additionally, a significance level, _$α$_, of five percent will be set. If the p-value is less than this threshold, then the null hypothesis can be rejected in favor of the alternative hypothesis.

```python
#Retrieve ozone values for each group
data_IV_wk_oz = data_IV[data_IV['Week/Weekend'] == 'Week']['Daily Max 8-hour Ozone Concentration'].values
data_IV_wknd_oz = data_IV[data_IV['Week/Weekend'] == 'Weekend']['Daily Max 8-hour Ozone Concentration'].values
#check # observations
display(len(data_IV_wk_oz), len(data_IV_wknd_oz))
```

```python
#Conduct hypothesis test
display(stats.ttest_ind(data_IV_wk_oz, data_IV_wknd_oz, alternative='two-sided'))

#Calculate the difference in average ozone levels on weekdays vs weekends
wk_wknd_avg_oz_diff = data_IV[data_IV['Week/Weekend'] == 'Week']['Daily Max 8-hour Ozone Concentration'].mean() - data_IV[
                        data_IV['Week/Weekend'] == 'Weekend']['Daily Max 8-hour Ozone Concentration'].mean()
display("Difference in mean ozone levels during the week & on the weekends in California, 2024: {} ppm.".format(
            np.round(wk_wknd_avg_oz_diff, 5)))
```

These results indicate that the probability of observing such a difference in ozone levels (or greater) in the main dataset is about 0.26% assuming that there there is no difference in average ozone levels between weekdays & weekends. In other words, if ozone levels were equivalent on weekdays & weekends in California in 2024, this test indicates that there is a 0.26% likelihood of observing a difference of 0.00043 ppm or more between the groups.  
Since the p-value is less than the established significance level of 0.05, the null hypothesis can be rejected. In other words, there is sufficient evidence to suggest that there was a meaningful difference in average ozone levels between weekdays & weekends in California in 2024.
- With this being said, the difference in average ozone levels between the two groups, shown above, is still very small; ~ 0.00043 ppm. Specifically, the average ozone levels observed during the week were about 0.00043 ppm higher than on the weekends.

These conclusions are similar to those observed in the original project (_Cleaning Data & the Skies I_), except that in the original version, there was not enough evidence to credibly assert that there was a meaningful difference in ozone levels between the two groups across the state in 2024.
